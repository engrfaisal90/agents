## Sidebar: Three concepts as a refresher

1. System Prompt: the part of the input to the LLM that describes the overall context of the conversation

2. Conversation History: the complete conversation so far

3. The illusion of memory: every message to an LLM is stateless. We pass in the complete conversation so far to give the illusion that it remembers what was said 30 seconds ago...

## Back to the main plot!

We have a LinkedIn profile in variable `linkedin`

We have a summary in variable `summary`

Let's construct a System Prompt..

```python
# Now lets Define a system prompt for the LLM. providing the details and cotext about the person we are representing and with the rules and context to follow.
system_prompt = f"""
# Define a system prompt for the LLM.
# Your role

You are a digital twin running on a website, chatting with visitors of the website.
You represent the person who's website you are on.
You answer questions related to their career, background, skills and experience.

Here are the details of the person you are representing:

{summary}

If asked, you explain clearly that you are an AI that is the digital twin of this person.

# Context

Here is a summary of the person's LinkedIn profile so that you can answer questions:

{linkedin}

# Rules

Engage with the user. Be professional and engaging, as if talking to a potential client or future employer who came across the website.
Avoid answering questions that are not related to the user's career, background, skills and experience;
steer the conversation back to professional topics.

Always stay in character as the digital twin of the person you are representing. Represent the person.

IMPORTANT: If you don't know the answer, say so. Never make up an answer.
If the user asks about something not in the context, say that you don't know.
"""
```

```python
# Display the system prompt.
display(Markdown(system_prompt))
```

```python
# Define a list of messages to send to the LLM.
messages = [
    # The system message is the context of the conversation.
    {"role": "system", "content": system_prompt},
    # The user message is the message from the user.
    {"role": "user", "content": "Hi - please tell me about yourself"},
]   
```

```python
#send the messages to the LLM.
response = openai.chat.completions.create(model=MODEL, messages=messages)
# Display the response from the LLM.
display(Markdown(response.choices[0].message.content))
```

```python
# Define a function named chat that accepts two parameters:
# message: The latest message or question submitted by the user.
# history: A list containing previous messages in the conversation.
def chat(message, history):
    # Reconstruct the conversation history using only the 'role' and 'content' fields of each message. 
    # This list comprehension iterates through every message (h) in the history list and creates a new dictionary containing its role and content.
    # This ensures that the conversation history follows the expected message structure.
    history = [{"role": h["role"], "content": h["content"]} for h in history]

    # Construct the complete messages list that will be sent to the language model.
    # 1. System message: Provides instructions about the context how the AI should behave.
    # 2. History: Includes previous messages in the conversation.
    # 3. User message: Contains the latest question or input.
    # The + operator combines these three lists into one list.

    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]

    #   #send the messages to the LLM.
    response = openai.chat.completions.create(model=MODEL, messages=messages)
    # return the LLM reponse
    return response.choices[0].message.content
```

```python
#lets call the above chat method now with new message and  and no history . The chat function already has the system prompt "conetx and rules" 
new_message="Please summarize who you are"
history=[]
chat(new_message, history)
```

```python
# Create a Gradio ChatInterface and connect it to the previously defined chat() function.

# gr.ChatInterface() automatically creates a chatbot interface with a message input box, conversation display, and submit functionality.

# chat: The Python function that processes user messages, sends them to the OpenAI API, and returns AI-generated responses.

# .launch(): Starts the Gradio application and makes the chatbot interface accessible through a web browser.

# inbrowser=True: Automatically opens the chatbot interface in the default web browser when the application starts.

gr.ChatInterface(chat).launch(inbrowser=True)
```

# And now - TOOLS!

Let's start with a function...

```python

# Create a tool to save users' email addresses in a text file
def record_email_tool(email):
    # Display the received email in the console
    print(f"Tool called to record an email: {email}")
    # Open emails.txt in append mode (create if it doesn't exist)
    with open("emails.txt", "a", encoding="utf-8") as f:
         # Save the email address on a new line
        f.write(email + "\n")
    # Return a confirmation message
    return "Email received"
```

```python
record_email_tool("test@abc.com")
```

## Step 1 - write some json to describe the tool


```python

# Define the tool specification that will be provided to the AI model
record_email_tool_json = {

    # Name of the tool (must match the Python function name)
    "name": "record_email_tool",

    # Explain when and why the AI should use this tool
    "description": "Use this tool to record that a user provided their email address",

    # Define the input parameters required by the tool
    "parameters": {
        "type": "object",

        # Specify the available input fields
        "properties": {
            "email": {
                "type": "string",
                "description": "The email address of this user"
            }
        },

        # Email is mandatory for executing this tool
        "required": ["email"],

        # Do not allow additional, undefined parameters
        "additionalProperties": False
    }
}
```

```python
# Create a list of tools available to the AI model
tools = [

    {
        # Specify that this tool is a function
        "type": "function",

        # Provide the function's name, description, and parameters
        "function": record_email_tool_json
    }
]
```

## Step 2 - a new chat() function

This is where we implement the tool call.

The reality is, it's a bit clunky. This is like seeing the ingredients of a fine recipe, and finding that the ingredients turn out to be quite ordinary.

Tool calling is an "if" statement. In this case, we're hardcoding everything to assume that the only tool is an email tool.

SIDENOTE: If you're thinking - but wait! I should be remembering this so I can do it myself! Then the key point is: this is what Agent Frameworks take care of for you. In practice, you'll likely never type this again yourself. We are shielded from these if statements by the Agent Framework. That's why they're often described as "abstraction layers".

```python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model=MODEL, messages=messages, tools=tools)
   
         
    if response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            tool_call = message.tool_calls[0]
            email = json.loads(tool_call.function.arguments).get("email")
            record_email_tool(email)
            messages.append(message)
            messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model=MODEL, messages=messages, tools=tools)
            
    return response.choices[0].message.content
```

```python
# new_message="Please summarize who you are"
# history=[]
# chat(new_message, history)
gr.ChatInterface(chat).launch(inbrowser=True)
```

## Step 3

Our first ever Agent Loop, done without an Agent Framework!

Changes:
1. Instead of always assuming there's only 1 tool call, iterate through the tools with a for loop
2. Changed from `if finish_reason=="tool_calls"` to `while finish_reason=="tool_calls"`

```python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model=MODEL, messages=messages, tools=tools)
    
    while response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            messages.append(message)
            for tool_call in message.tool_calls:
                email = json.loads(tool_call.function.arguments).get("email")
                record_email_tool(email)
                messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
            
    return response.choices[0].message.content
```

```python
gr.ChatInterface(chat).launch(inbrowser=True)
```
